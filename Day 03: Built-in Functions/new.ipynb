1️⃣ Arithmetic + Precedence

```python
a = 8
b = 3
c = 2

print(a + b * c - a // b)
```

✅ Step-by-step (operator precedence)

Order:

* and //
+ and -
Step 1: Multiplication

```python
b * c = 3 * 2 = 6
```

Step 2: Floor Division

```python
a // b = 8 // 3 = 2
```

👉 // means floor division → removes decimal → 2.66 → 2

Step 3: Final expression

```python
a + 6 - 2 = 8 + 6 - 2 = 12
```

Output:

```python
12
```
